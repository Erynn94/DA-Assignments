```python
# Visualize price against carat (carat on x, price on y)
diamonds = sns.load_dataset('diamonds')
sns.scatterplot(data=diamonds, x='carat', y='price')
plt.title('Price vs. Carat')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.show()
```

```python
# Now with a regression line
sns.lmplot(data=diamonds, x='carat', y='price', line_kws={'color': 'red'})
plt.title('Price vs. Carat (with Regression Line)')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.show()
```

```python
# With regression lines for each color
sns.lmplot(data=diamonds, x='carat', y='price', hue='color')
plt.title('Price vs. Carat (by Color)')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.show()
```

```python
# Remove missings
oil_clean = oil.dropna()
oil_clean.shape
```

```python
# Convert date column to datetime
oil_clean.loc[:, 'date'] = pd.to_datetime(oil_clean['date'])
```

```python
# Set date column as index
oil_clean.set_index('date', inplace=True)
oil_clean.head()
```

```python
# Create a series with 700 observations
series_700 = pd.Series(np.random.randn(700))
series_700.head()
```

```python
# Create a series with only prices over $50 (just for fun, we won't use this)
prices_over_50 = oil_clean[oil_clean['dcoilwtico'] > 50]['dcoilwtico']
prices_over_50.head()
```

```python
# Generate a line graph
plt.plot(oil_clean['dcoilwtico'])
plt.title('WTI Crude Oil Prices Over Time')
plt.xlabel('Date')
plt.ylabel('Price (US$)')
plt.show()
```

```python
# Generate a decomposition graph
from statsmodels.tsa.seasonal import seasonal_decompose

result = seasonal_decompose(oil_clean['dcoilwtico'], model='additive', period=30)
result.plot()
plt.show()
```
